```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import sounddevice as sd

t = np.arange(0, 5, 0.00001) # 0.00001 = 0.01 ms

def x1(t):
    return np.sin(np.pi*2/0.01*t)

def x2(t):
    return np.sin(np.pi*2*1000*t-np.pi/4)

# Create a new figure
fig , ax = plt.subplots( )
# Plot the curve with x−values ’t’ and y−values ’x’ into

# the figure ’ax’ with the legendentry ’My Signal ’
ax.plot(t, x1(t), label="Signal 1")
ax.plot(t, x2(t), label="Signal 2")

#Visar signalernas namn? (signal 1 och signal 2)
ax.legend()

#labels
ax.set_xlabel("Tid (s)")
ax.set_ylabel("x(t)")

# Vi kör i 10ms
ax.set_xlim(0, 0.01)
ax.set_ylim(-1.5, 1.5)

ax.grid()

plt.show()

fs = 100000

#sd.play(x2(t), fs , blocking=True)
sd.play(x2(t), fs , blocking=True)
```

När vi spelar upp den andra sinus-signalen x2, hör vi snarare ett högre ljus, som låter mer likt en en-tonig alarmsignal. Det låter som en signal som skulle kunna komma ifrån en dörr som öppnats som endast var en nödsignal, eller en signal från en telefon som har blivit överhettad.
