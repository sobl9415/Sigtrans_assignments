```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import sounddevice as sd

dt = 0.00001  # 0.01 ms tidssteg
t = np.arange(0, 0.01, dt) # 0.00001 = 0.01 ms
t_audio = np.arange(0, 5, dt) # lägre sluttid för ljuduppspelningen

def x1(t):
    return np.sin(np.pi*2/0.01*t)

def x2(t):
    return np.sin(np.pi*2*1000*t-np.pi/4)

# Create a new figure
fig , ax = plt.subplots( )
# Plot the curve with x−values ’t’ and y−values ’x’ into

ax.plot(t*1000, x1(t), label="Signal 1") 
ax.plot(t*1000, x2(t), label="Signal 2")

# visar signalernas namn, (signal 1 och signal 2)
ax.legend()

# labels
ax.set_xlabel("Tid (ms)")
ax.set_ylabel("x(t)")

# Vi kör i 10ms
ax.set_xlim(0, 10)
ax.set_ylim(-1.5, 1.5)

ax.grid()

plt.show()

fs = 100000

#sd.play(x1(t_audio), fs , blocking=True)
#sd.play(x2(t_audio), fs , blocking=True)
```
